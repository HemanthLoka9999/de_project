```python
def inspect_df(df, name, deep=False):
    """
    Inspect a DataFrame in a manageable way.
    
    Parameters:
    - df: DataFrame to inspect
    - name: Name of the DataFrame
    - deep: If True, shows full details; if False, shows essentials only
    """
    print(f"\n--- {name} ---")
    
    # Essentials
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Missing values (first 10 columns):\n", df.isnull().sum().head(10))
    
    if deep:
        # More detailed inspection
        print("Data types:\n", df.dtypes)
        print("Duplicate rows:", df.duplicated().sum())
        
        categorical_cols = df.select_dtypes(include='object').columns.tolist()
        numeric_cols = df.select_dtypes(include='number').columns.tolist()
        date_cols = [col for col in df.columns if 'date' in col or 'timestamp' in col]
        head_data = df.head(5)

        print("Categorical columns:", categorical_cols)
        print("Numeric columns:", numeric_cols)
        print("Date columns:", date_cols)
        print("Head data:", head_data)

    
    print("-" * 50)

# Example usage
inspect_df(df_customers, "Customers")          # Essentials only
inspect_df(df_orders, "Orders", deep=True)     # Full inspection
```
